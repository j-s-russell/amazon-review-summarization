# Results and Analysis

This notebook loads our results, displays ROUGE scores and BERTScores in tables for each summary type, and provides sample summaries for comparison.

```python
# Import necessary libraries

import json
import pandas as pd
import textwrap
```

## Performance Analysis - ROUGE

```python
# Load ROUGE results

file_paths = {
    "kl_sum_rouge_results": "../results/kl_sum_rouge_results.json",
    "lexrank_rouge_results": "../results/lexrank_rouge_results.json",
    "bart_rouge_results": "../results/bart_rouge_results.json",
    "t5_rouge_results": "../results/t5_rouge_results.json"
}

# Load each JSON file into a dictionary
kl_sum_rouge_results = json.load(open(file_paths["kl_sum_rouge_results"], "r"))
lexrank_rouge_results = json.load(open(file_paths["lexrank_rouge_results"], "r"))
bart_rouge_results = json.load(open(file_paths["bart_rouge_results"], "r"))
t5_rouge_results = json.load(open(file_paths["t5_rouge_results"], "r"))
```

```python
all_results = {
    "KL-Sum": kl_sum_rouge_results,
    "LexRank": lexrank_rouge_results,
    "BART": bart_rouge_results,
    "T5": t5_rouge_results
}

# Initialize empty dictionaries to store per-section data
verdict_rows = {}
pros_rows = {}
cons_rows = {}

# Extract ROUGE scores by section and model
for model_name, results in all_results.items():
    verdict_rows[model_name] = results["verdict"]
    pros_rows[model_name] = results["pros"]
    cons_rows[model_name] = results["cons"]

# Convert to DataFrames
verdict_df = pd.DataFrame.from_dict(verdict_rows, orient="index").round(4)
pros_df = pd.DataFrame.from_dict(pros_rows, orient="index").round(4)
cons_df = pd.DataFrame.from_dict(cons_rows, orient="index").round(4)

# Optional: Sort rows alphabetically by model name
verdict_df = verdict_df.sort_index()
pros_df = pros_df.sort_index()
cons_df = cons_df.sort_index()

# Display the tables
print("Verdict ROUGE Scores:")
display(verdict_df)

print("\nPros ROUGE Scores:")
display(pros_df)

print("\nCons ROUGE Scores:")
display(cons_df)
```

## Performance Analysis - BERTScore

```python
# Load BERTScore results

file_paths = {
    "kl_sum_bertscore_results": "../results/kl_sum_bertscore_results.json",
    "lexrank_bertscore_results": "../results/lexrank_bertscore_results.json",
    "bart_bertscore_results": "../results/results/bart_bertscore_results.json",
    "t5_bertscore_results": "../results/t5_bertscore_results.json"
}

# Load each JSON file into a dictionary
kl_sum_bertscore_results = json.load(open(file_paths["kl_sum_bertscore_results"], "r"))
lexrank_bertscore_results = json.load(open(file_paths["lexrank_bertscore_results"], "r"))
bart_bertscore_results = json.load(open(file_paths["bart_bertscore_results"], "r"))
t5_bertscore_results = json.load(open(file_paths["t5_bertscore_results"], "r"))
```

```python
bertscore_results = {
    "KL-Sum": kl_sum_bertscore_results,
    "LexRank": lexrank_bertscore_results,
    "BART": bart_bertscore_results,
    "T5": t5_bertscore_results
}

# Create section-specific dictionaries
verdict_bert = {}
pros_bert = {}
cons_bert = {}

# Extract scores by section and model
for model_name, scores in bertscore_results.items():
    verdict_bert[model_name] = scores["verdict"]
    pros_bert[model_name] = scores["pros"]
    cons_bert[model_name] = scores["cons"]

# Convert to DataFrames and round for display
verdict_bert_df = pd.DataFrame.from_dict(verdict_bert, orient="index").round(4)
pros_bert_df = pd.DataFrame.from_dict(pros_bert, orient="index").round(4)
cons_bert_df = pd.DataFrame.from_dict(cons_bert, orient="index").round(4)

# Optional: Sort for consistent order
verdict_bert_df = verdict_bert_df.sort_index()
pros_bert_df = pros_bert_df.sort_index()
cons_bert_df = cons_bert_df.sort_index()

# Display in Jupyter Notebook
print("Verdict BERTScore:")
display(verdict_bert_df)

print("\nPros BERTScore:")
display(pros_bert_df)

print("\nCons BERTScore:")
display(cons_bert_df)
```
